### 2) 텍스트 전처리

(1-1) **불필요한 문자 제거**                   
    (예: `#`, `\t\t\t\t\t\t\t\t\t`, 빈 문자열('') 등)               

(1-2) **[] 대괄호 제거**                   

(1-3) **공백 대체 후 단어 분리**               
    * 단어 사이 `/`, `.`, `+`를 공백으로 변환 후  구분자(,)로 분리                 
    (예: `휴식/힐링` → '휴식 힐링' `휴식`, `힐링`)                   
    (예: `비.눈` → '비 눈' ->  `비`, `눈`)                  
    (예: `실내+실외` → '실내 실외' -> `실내`, `실외`)              
    ※ 단, `4.3사건`과 같은 고유명사는 변환하지 않도록 예외 처리                    
                

```python
df['merged_tag_cleaned'] = (
    df['merged_tag'].astype(str)
      .str.replace(r"[\\t]", "", regex=True)      # '\t' 반복 → 삭제
      .str.replace(r"[#\[\]'\"]", "", regex=True)  # '#', '[]', 따옴표 삭제
      .str.replace(r"[\/+]", ",", regex=True)      # '/', '+' → ,로 변경
      .str.replace(r"(?<!\d)\.(?!\d)", ",", regex=True)  # 숫자가 아닌 점 → ,로 변경
      .str.replace(r"['']", '', regex=True)        # 빈값 → 삭제
      .str.strip()                      # 양쪽 공백 제거
)
```
